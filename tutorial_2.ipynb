#### In a 6G network, massive MIMO (Multiple Input Multiple Output) technology employs arrays of antennas to enhance capacity and coverage. Each patch antenna operates at 60 GHz with an effective area of 0.001 $m^2$. Based on this scenario, please answer the following:

#### a) What is the gain of a single antenna element, expressed in dBi?

```python
3e8/60e9
```

```python
import math
pi = math.pi
log10 = math.log10
# So - Gain of single element in dBi
f = 60e9
Ae = 0.001
wl = 3e8/f # 0.005 m = 0.5cm

G_linear = 4*pi*Ae/wl**2
G_dbi = 10*log10(G_linear)
print("G_linear: ", G_linear)
print("G_dbi: ", G_dbi, 'dBi')
```

#### b) If the array consists of 128 antennas and all signals add constructively, what would be the gain of the combined array?

> The formula for calculating constructive interference is as follows:
$$G^{dBi} = G_{single}^{dBi} + 10*\log_{10}(N)$$

```python
# So in my case:
G_total = G_dbi + 10*log10(128)
print("G_total: ", G_total, 'dBi')
```
